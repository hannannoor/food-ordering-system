```python
menu_items=["Cheeseburger","Pepperoni Pizza","Chicken Nuggets","Fries","Coca Cola","Carbonara","Bolognese",]
menu_prices=[7.00,15.00,6.00,4.50,1.50,8.50,10.00]
number_of_menu_items= len(menu_items)


#contains list of items 
current_order_items =[]
# contains list of quantities of items
current_order_quantities=[]

# the max number of items a user can add at one time
max_items=999

# this is a nested array that includes  [order_number, list_of_items, list_of_quantities, list_of_item_totals, subtotal_string]
order_history_record=[]

# contains the promo codes
promo_codes=["STUDENT10","NSF20","SNR20","STAFF30","NATIONALDAY20"]
#contains the discount percentages of the promo codes
discount_amounts=[10,20,20,30,20]
# contains a list of used promo codes, once a promo code is used it is 
used_promo_codes=[]

#contains list of items in the points menu
points_menu_items=["Coca Cola","Vanilla Ice Cream","Cheese Fries","Choco-Cone"]
# containts list of cost(in points) of the items in the points menu
points_menu_cost=[50,100,200,150]
# number of points user has
points=[0]



def number_validation(s,minimum,maximum):
    # Checks if input is a positive integer and in the range minimum-maximum
    if s=="":
        return False
    for character in s:
        # if any "character" in string entered is not between "0" and "9" returns False
        if character not in "0123456789":
            return False
    if int(s)<minimum or int(s)>maximum:
        return False
    return True
    
def format_currency(amount):
    # converts float/integer(2.352) into a string like $2.35
    # the +0.005 is so that it rounds down/up correctly
    cents=int(amount*100+0.005)
    dollars=cents//100
    remaining_cents=cents%100
    # if cents less than 10 adds a 0 before , i.e. if  amount is 10 dollars and 9 cents makes it 10.09 instead of 10.9
    if remaining_cents < 10:
        cents_str="0"+ str(remaining_cents)
    else:
        cents_str = str(remaining_cents)
    amount_str = "$"+str(dollars)+"."+cents_str
    return amount_str

def find_index(item,arr):
    # finds the index of the item ie if its cheeseburger returns 0 as index of cheeseburger is 0 in menu items
    idx=0
    for i in arr:
        if i == item:
            return idx
        else:
            idx+=1

def clear_arr(arr):
    # clears the array by converting the last element in the array to empty space then uses .remove to remove empty space and loops until the arrary is empty
     while len(arr) > 0:
        arr[len(arr) - 1] = ""
        arr.remove("")
    
def calculate_total():
    total= 0.0
    # calculates the total price of all the items so far
    for i in range(0,len(current_order_items)):
        index_of_item=find_index(current_order_items[i],menu_items)
        total+=(menu_prices[index_of_item]*current_order_quantities[i])
    return (total)
        
def view_menu():
    print("=== Menu ===")
    # shows menu , repeats loop on showing each menu item and its price , for how many menu items there are
    for i in range(0,number_of_menu_items):
        name=menu_items[i]
        price=menu_prices[i]
        print(str(i+1)+") "+name + ": " + format_currency(price))
        
def add_item():
    view_menu()
    print(" ")
    valid_input=False
    # loops until a valid input is entered or user enters 0 to exit
    while valid_input==False:
        menu_prompt="Enter the number of the item you would like to order (1-"+str(number_of_menu_items)+") or enter 0 to Exit"+": "
        choice=input(menu_prompt)
        if number_validation(choice,0,number_of_menu_items) == False:
            print("Invalid Input, Enter a number in the range","1-",number_of_menu_items)
        else:
            choice=int(choice)
            if choice==0:
                return
            valid_input=True
            
    valid_input=False
    while valid_input==False:
        qty_input = input("Enter quantity or enter 0 to Exit: ")
        if number_validation(qty_input,0,max_items) == False:
            print("Invalid quantity. Please type a whole number from 0-",max_items)
        else:
            quantity=int(qty_input)
            valid_input=True
            if quantity==0:
                return
    item_name=menu_items[choice-1]
    # if item is already added just adds the new quantity to the old quantity.
    # if item is not already added then appends the item name to the current_order_items and the quantity to
    # current_order_quantities
    if item_name in current_order_items:
        idx=find_index(item_name,current_order_items)
        current_order_quantities[idx]=current_order_quantities[idx]+quantity
    else:
        current_order_items.append(menu_items[choice-1])
        current_order_quantities.append(quantity)
    
    print(str(quantity) + " x " + item_name + " added to your current order.")
    
    
def display_current_order():
    if len(current_order_items) == 0:
        print("You have not added anything yet.")
        return

    print("=== Current Order ===")
    total = 0
    num=1
    for i in range(len(current_order_items)):
        item_name = current_order_items[i]
        quantity = current_order_quantities[i]
        unit_price = menu_prices[find_index(item_name,menu_items)]
        item_total=unit_price*quantity

        print(str(num)+")Item: " + item_name)
        print("  Quantity: " + str(quantity))
        print("  Unit price: " + format_currency(unit_price))
        print("  Item total: " + format_currency(item_total))
        num+=1

    print("Order Subtotal: " + format_currency(calculate_total()))
    print("----------------------")

        
def update_item():
    #if theres no items in the current order gives error message and exits the function
    if len(current_order_items) == 0:
        print("Your order is empty. Nothing to update.")
        return
    # shows the current order which includes numbering
    display_current_order()
    print("""
     """)
    valid_input=False
    # asks`users which item they want to edit the quantity/remove 
    while valid_input==False:
        choice = input("Enter the number of item you want to update or remove or enter 0 to exit: ")
        if number_validation(choice,0,len(current_order_items))==False:
            print("Invalid input, enter a number from 0-"+str(len(current_order_items)))
        else:
            # converts the input to a integer and allows user to exit
            choice=int(choice)
            valid_input=True
            if choice==0:
                return
    valid_input=False
    # asks users to what the new quantity of the is , and 0 if they want to remove the item
    while valid_input == False:
        new_qty_input = input("Enter new quantity (0 to remove): ")
        if not number_validation(new_qty_input,0,max_items):
            print("Invalid input. Please type a whole number from 0 to",max_items)
            continue
        new_qty = int(new_qty_input)
        valid_input=True
        #  removes the item if 0 is entred
        if new_qty == 0:
            current_order_items.remove(menu_items[choice-1])
            current_order_quantities.remove(menu_items[choice-1])
            print(menu_items[choice-1] + " has been removed from your order.")
        # updates the quantity if 0 is not entered
        else:
            current_order_quantities[choice-1] = new_qty
            print(current_order_items[choice-1] + " quantity updated to " + str(new_qty) + ".")
 
def applypromo(total):
    # loop to obtain promo code, users can also exit by entering 0
    while True:
        # possible promo codes STUDENT10,NSF20,SNR20,STAFF30,NATIONALDAY20
        disc_code=input("Enter a Promo Code, or enter 0 if you do not have a Promo Code: ").upper()
        if disc_code=="0":
            return total,0
        # checks if code has already been used and if so gives an error message and continues the loop
        elif disc_code in used_promo_codes:
            print("Promo Code has already been used")
            continue
        elif disc_code in promo_codes:
            break
        else:
            print("Invalid Promo Code")
    
    idx=find_index(disc_code,promo_codes)
    discount_percentage=discount_amounts[idx]
    #calculates new subtotal after applying discount
    subtotal=(total*(100-discount_percentage))/100
    # prints conformation that discount code has been used
    print("Promo Code '"+disc_code+"' applied for "+str(discount_percentage)+"% discount")
    
    # adds the used promo code to the used_promo_code array so that later it cant be reused
    used_promo_codes.append(disc_code)
    # returns the new total and the discount applied
    return subtotal,discount_percentage


def checkout():
    if len(current_order_items)==0:
        print("Unable to checkout because your order is empty")
        return
    # displays order 
    print("=== Checkout ===")
    display_current_order()
    
    # checks if user wants to apply a promo code 
    subtotal,discount_applied=applypromo(calculate_total())
    total_str=format_currency(subtotal)
    print("Total amount to pay:",total_str)
    
    
    # checks  for conformation
    confirm = input("Type 'Yes' to confirm, anything else to cancel: ").lower()
    print(" ")
    if confirm != "y" and confirm  != "yes":
        print("Checkout cancelled")
        return
    
    # adds 10 points for every dollar spend(only whole numbers no decimals for points)
    points_added=subtotal*10
    current_points=points[0]
    total_points=int(current_points+points_added)
    
    # updates the integer in the arr points to the new total points
    points[0]=total_points
    
    print(points[0],"points added to balance (10 points for every dollar spent)")
    print("Points can be redeemed for food in the Redeem Points section")
    
    # stores current order data to order_history arrays 

    saved_items = []
    saved_quantities = []
    saved_item_totals = []

    for i in range(len(current_order_items)):
        item = current_order_items[i]
        qty = current_order_quantities[i]
        price = menu_prices[find_index(item, menu_items)]
        item_total = price * qty

        saved_items.append(item)
        saved_quantities.append(qty)
        saved_item_totals.append(item_total)
        
    order_number = len(order_history_record) + 1
    single_order = [
        order_number,
        saved_items,
        saved_quantities,
        saved_item_totals,
        total_str,
        discount_applied
    ]
    
    order_history_record.append(single_order)
    
    print("Order confirmed and saved as Order #" + str(order_number) + ".")
    
    #clears current order
    clear_arr(current_order_items)
    clear_arr(current_order_quantities)
    
    
    
        
def view_order_history():
    # if there arnt any past orders stops the function and gives error message
    if len(order_history_record)==0:
        print("No Orders completed yet.")
        return
    print("--- Order History ---")
    for order in order_history_record:
        order_number= order[0]
        items= order[1]
        quantities=order[2]
        item_totals = order[3]
        subtotal = order[4]
        discount_applied= order[5]
        
        print("Order #"+str(order_number)+"")
        print("           ")
        for i in range(len(items)):
            item_name=items[i]
            price=menu_prices[find_index(item_name, menu_items)]
            print("  Item: " + item_name)
            print("  Quantity: " + str(quantities[i]))
            print("  Unit price: " + format_currency(price))
            print("  Item total: " + format_currency(item_totals[i]))
            print("")
        print("Discount Applied: "+ str(discount_applied) +"%")
        print("Subtotal: "+subtotal)   
        print("------------")   
        
# my 2nd extra feature is , when user purchases food they can gain points, and can redeem points to buy extra food 

#this function shows the points menu and the number of points
def points_menu():
    print("=== Points Menu ===")
    for i in range(len(points_menu_items)):
        name=points_menu_items[i]
        price=points_menu_cost[i]
        print(str(i+1)+") "+name + ": " + str(price)+" points")
    print("Points Balance: "+str(points[0]))

#this function allows users to redeem their points to get food
def redeem_points():
    points_menu()
    # asks users what item they want to get
    valid_input=False
    while valid_input==False:
        points_menu_prompt="Enter the number of the item you would like to get (1-"+str(len(points_menu_items))+") or enter 0 to Exit"+": "
        choice=input(points_menu_prompt)
        if number_validation(choice,0,len(points_menu_items)) == False:
            print("Invalid Input, Enter a number in the range","1-"+str(points_menu_items))
        else:
            choice=int(choice)
            if choice==0:
                return
            valid_input=True
    # asks users what quantity of items they want to get
    valid_input=False
    while valid_input==False:
        points_qty_input = input("Enter quantity or enter 0 to Exit: ")
        if number_validation(points_qty_input,0,max_items) == False:
            print("Invalid quantity. Please type a whole number between 0 to",max_items)
        else:
            item_quantity=int(points_qty_input)
            valid_input=True
            if item_quantity==0:
                return
        
    # assigns variables for item_name and item cost and total points cost
    idx=choice-1
    item_name=points_menu_items[idx]
    item_cost=points_menu_cost[idx]
    total_points_cost=item_cost*item_quantity
    # gives error message if user does not have enough points
    if total_points_cost>points[0]:
        print("Insufficient Balance, Buy Food to gain points")
        return
    
    print(str(item_quantity) + " x " + item_name + " purchased for " + str(total_points_cost)+" points")
    # adjusting total points balance
    points[0]-= total_points_cost
    print ("Remaining Points Balance : "+str(points[0]))
        
def main():
    while True:
        print(" ")
        print("=== Python Food Cafe ===")
        print("1. View Menu")
        print("2. Add Item to Order")
        print("3. Update or Remove Item")
        print("4. Display Current Order")
        print("5. Checkout")
        print("6. View Order History")
        print("7. View Points Menu")
        print("8. Redeem Points")
        print("9. Exit")
        choice = input("Enter your choice (1-7): ")
        if choice == "1":
            view_menu()
        elif choice == "2":
            add_item()
        elif choice == "3":
            update_item()
        elif choice == "4":
            display_current_order()
        elif choice == "5":
            checkout()
        elif choice == "6":
            view_order_history()
        elif choice == "7":
            points_menu()
        elif choice == "8":
            redeem_points()
        elif choice == "9":
            print("Exiting...")
            break
                
        else:
            print("That is not a valid option. Enter 1 to 7.")

                 
main()
    
```
